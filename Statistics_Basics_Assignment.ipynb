Statistics Basics

Q1.What is statistics, and why is it important

Statistics is the science of collecting, analyzing, interpreting, and presenting data; it's important for informed decision-making.

Q2. What are the two main types of statistics

The two main types are descriptive statistics and inferential statistics.

Q3. What are descriptive statistics

Descriptive statistics summarize and describe the main features of a dataset (e.g., mean, median, mode).

Q4. What is inferential statistics

Inferential statistics use sample data to make generalizations or predictions about a population.

Q5. What is sampling in statistics

Sampling is the process of selecting a subset from a population to represent the whole.

Q6. What are the different types of sampling methods

Types of sampling methods include random, stratified, systematic, cluster, and convenience sampling.

Q7. What is the difference between random and non-random sampling

Random sampling gives each member equal chance; non-random sampling doesn’t and may introduce bias.

Q8. Define and give examples of qualitative and quantitative data

Qualitative data are non-numerical (e.g., colors); quantitative data are numerical (e.g., height, age).

Q9. What are the different types of data in statistics

Types of data: nominal, ordinal, interval, ratio (based on level of measurement).

Q10. Explain nominal, ordinal, interval, and ratio levels of measurement.

Nominal: categories (gender); Ordinal: order matters (rank); Interval: equal intervals, no true zero (temperature); Ratio: equal intervals, true zero (weight).

Q11. What is the measure of central tendency

Measures of central tendency locate the center of a dataset (mean, median, mode).

Q12. Define mean, median, and mode

Mean is the average, median is the middle value, mode is the most frequent value.

Q13. What is the significance of the measure of central tendency

Central tendency helps in understanding the typical or average value in data.

Q14. What is variance, and how is it calculated

Variance measures the average squared deviation from the mean; calculated by averaging squared differences.

Q15. What is standard deviation, and why is it important

Standard deviation is the square root of variance; it shows how spread out the data is.

Q16. Define and explain the term range in statistics

Range is the difference between the maximum and minimum values in a dataset.

Q17. What is the difference between variance and standard deviation

Variance uses squared units; standard deviation is in original units and more interpretable.

Q18. What is skewness in a dataset

Skewness measures the asymmetry of a data distribution.

Q19. What does it mean if a dataset is positively or negatively skewed

Positive skew: tail on right; Negative skew: tail on left.

Q20. Define and explain kurtosis

Kurtosis describes the "tailedness" of a distribution—how extreme values are.

Q21. What is the purpose of covariance

Covariance shows the direction of the relationship between two variables.

Q22. What does correlation measure in statistics

Correlation measures both the strength and direction of the relationship.

Q23.  What is the difference between covariance and correlation

Covariance shows direction only; correlation standardizes it to a value between -1 and 1.

Q24. What are some real-world applications of statistics?

Applications include business forecasting, healthcare analysis, quality control, and social science research.

How do you calculate the mean, median, and mode of a dataset

```python
import statistics

data = [12, 15, 21, 15, 19, 15, 22]

mean = statistics.mean(data)
median = statistics.median(data)
mode = statistics.mode(data)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

Implement sampling techniques like random sampling and stratified sampling

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'value': np.arange(100)})
# Random Sampling
random_sample = df.sample(n=10)
# Stratified Sampling
df['strata'] = pd.cut(df['value'], bins=5, labels=False)
stratified_sample = df.groupby('strata').apply(lambda x: x.sample(2)).reset_index(drop=True)
```
